```python
import pandas as pd
import numpy as np

df = pd.read_csv('options_merged_spx.csv', sep=';')

# 1. Vue d'ensemble
print(f"Dimensions du dataset : {df.shape[0]:,} lignes × {df.shape[1]} colonnes")

# 2. Valeurs manquantes par colonne
missing = df.isnull().sum()
pct     = (missing / len(df) * 100).round(4)
tableau = pd.DataFrame({
    'Valeurs manquantes' : missing,
    'Pourcentage (%)'    : pct,
    'Complétude (%)'     : (100 - pct).round(4)
})
print("\n=== Valeurs manquantes par colonne ===")
print(tableau[tableau['Valeurs manquantes'] > 0])
if tableau['Valeurs manquantes'].sum() == 0:
    print("→ Aucune valeur manquante détectée.")

# 3. Vérifications de cohérence métier
print("\n=== Vérifications de cohérence ===")
print(f"iv <= 0       : {(df['iv'] <= 0).sum()} lignes")
print(f"iv > 5 (>500%): {(df['iv'] > 5).sum()} lignes")
print(f"tenor_d <= 0  : {(df['tenor_d'] <= 0).sum()} lignes")
print(f"volume < 0    : {(df['volume'] < 0).sum()} lignes")
print(f"strike <= 0   : {(df['strike'] <= 0).sum()} lignes")

# 4. Statistiques iv
print("\n=== Distribution de la variable cible iv ===")
print(df['iv'].describe().round(4))
```
